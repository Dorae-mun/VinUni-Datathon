```python
import sys
import pandas as pd
sys.path.append("../src")

import data_loader
```

```python
customers = data_loader.load_customers()
orders = data_loader.load_orders()
order_items = data_loader.load_order_items()
```

```python
df = (
    orders
    .merge(customers, on="customer_id", how="left")
    .merge(order_items, on="order_id", how="left")
)
```

```python
df.columns
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
snapshot_date = df["order_date"].max() + pd.Timedelta(days=1)
```

```python
rfm = df.groupby("customer_id").agg({
    "order_date": lambda x: (snapshot_date - x.max()).days,  # Recency
    "order_id": "nunique",                                   # Frequency
    "line_revenue": "sum"                                     # Monetary
}).reset_index()
```

```python
rfm.columns = ["customer_id", "Recency", "Frequency", "Monetary"]
```

# Chia rule

```python
rfm["R_group"] = pd.cut(
    rfm["Recency"],
    bins=[-1,365,1000,10000],
    labels=["Recent","Warm","Old"]
)
```

```python
rfm["F_group"] = pd.cut(
    rfm["Frequency"],
    bins=[0,1,5,1000],
    labels=["One-time","Occasional","Loyal"]
)
```

```python
rfm["M_group"] = pd.cut(
    rfm["Monetary"],
    bins=[0,90000,300000,10000000],
    labels=["Low","Medium","High"]
)
```

```python
def segment(row):

    if row["R_group"]=="Recent" and row["F_group"]=="Loyal" and row["M_group"]=="High":
        return "Champions"

    if row["F_group"]=="Loyal" and row["M_group"]!="Low":
        return "Loyal customers"

    if row["R_group"]=="Old":
        return "Churn risk"

    return "Regular customers"
```

```python
rfm["segment"] = rfm.apply(segment, axis=1)
```

```python
rfm["segment"].value_counts()
```

```python
rfm.groupby("segment").agg({
    "Recency":"mean",
    "Frequency":"mean",
    "Monetary":"mean",
    "customer_id":"count"
})
```
